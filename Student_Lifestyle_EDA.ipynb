# Student Lifestyle & Academic Performance — Exploratory Data Analysis

## Objective

The objective of this project is to perform Exploratory Data Analysis (EDA) on student lifestyle and academic performance data.

The analysis will explore study habits, sleep patterns, social time, physical activity, stress levels, and their relationship with students' academic performance (CGPA).

The project will focus on understanding the dataset, cleaning the data, analysing individual variables, studying relationships between variables, identifying outliers, creating meaningful visualizations, and communicating important insights from the data.


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully!")
import pandas as pd
df = pd.read_csv("student_lifestyle_dataset.csv")
print("=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)
print("Rows and Columns:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())
print("=" * 70)
print("STEP 4 - DATASET OVERVIEW")
print("=" * 70)
print("\n1. FIRST 5 ROWS")
display(df.head())
print("\n2. LAST 5 ROWS")
display(df.tail())
print("\n3. DATASET SHAPE")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\n4. DATASET INFORMATION")
df.info()
print("\n5. STATISTICAL SUMMARY")
display(df.describe(include="all"))
print("\n6. DATA TYPES")
display(df.dtypes)
print("\n7. UNIQUE VALUES")

for column in df.columns:
    print(f"\n{column}:")
    print("Unique values:", df[column].nunique())
    
    if df[column].nunique() <= 10:
        print("Values:", df[column].unique())
        
print("\n" + "=" * 70)
print("DATASET OVERVIEW COMPLETED")
print("=" * 70)
print("=" * 70)
print("STEP 5 - DATA CLEANING")
print("=" * 70)
print("\n1. MISSING VALUES")
missing_values = df.isnull().sum()
display(missing_values.to_frame(name="Missing_Values"))
print("\n2. DUPLICATE ROWS")
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
print("\n3. DATA TYPES")
display(df.dtypes)
print("\n4. UNIQUE VALUES")

for column in df.columns:
    print(f"{column}: {df[column].nunique()} unique values")

print("\n5. NUMERICAL COLUMNS")

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()
print(numerical_columns)
print("\n6. CATEGORICAL COLUMNS")

categorical_columns = df.select_dtypes(include="object").columns.tolist()

print(categorical_columns)

if duplicate_count > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("\nDuplicates removed successfully.")
else:
    print("\nNo duplicate rows found. No rows removed.")

print("\n7. FINAL DATASET SHAPE")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])


print("\n" + "=" * 70)
print("DATA CLEANING CHECK COMPLETED")
print("=" * 70)
print("=" * 70)
print("STEP 6 - UNIVARIATE ANALYSIS")
print("=" * 70)

numerical_columns = df.select_dtypes(include=np.number).columns.tolist()

# Student_ID is an identifier, so exclude it from analysis
if "Student_ID" in numerical_columns:
    numerical_columns.remove("Student_ID")

# Identify categorical columns
categorical_columns = df.select_dtypes(include="object").columns.tolist()

print("\nNumerical Variables:")
for column in numerical_columns:
    print("-", column)

print("\nCategorical Variables:")
for column in categorical_columns:
    print("-", column)

numerical_columns = [
    "Study_Hours_Per_Day",
    "Extracurricular_Hours_Per_Day",
    "Sleep_Hours_Per_Day",
    "Social_Hours_Per_Day",
    "Physical_Activity_Hours_Per_Day",
    "GPA"
]

for column in numerical_columns:
    
    plt.figure(figsize=(8, 5))
    
    sns.histplot(
        data=df,
        x=column,
        kde=True
    )
    
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
    

for column in numerical_columns:
    
    plt.figure(figsize=(8, 4))
    
    sns.boxplot(
        data=df,
        x=column
    )
    
    plt.title(f"Boxplot of {column}")
    plt.xlabel(column)
    plt.tight_layout()
    plt.show()
    

print("=" * 70)
print("STRESS LEVEL - FREQUENCY ANALYSIS")
print("=" * 70)


stress_counts = df["Stress_Level"].value_counts()

print("\nStress Level Frequency:")
display(stress_counts.to_frame(name="Number_of_Students"))



plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Stress_Level",
    order=df["Stress_Level"].value_counts().index
)

plt.title("Distribution of Student Stress Levels")
plt.xlabel("Stress Level")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()


stress_percentage = (
    df["Stress_Level"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nStress Level Percentage:")
display(stress_percentage.to_frame(name="Percentage"))
```

 # 7.CORRELATION WITH GPA

```python
print("\nCorrelation of Numerical Variables with GPA:")

gpa_correlations = (
    df[numerical_columns]
    .corr()["GPA"]
    .sort_values(ascending=False)
)

display(gpa_correlations.to_frame(name="Correlation_with_GPA"))


print("\n" + "=" * 70)
print("BIVARIATE ANALYSIS COMPLETED")
print("=" * 70)
```

```python
print("=" * 70)
print("STEP 8 - MULTIVARIATE ANALYSIS")
print("=" * 70)
print("\n8A - CORRELATION HEATMAP")

correlation_matrix = df[numerical_columns].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Student Lifestyle Variables")
plt.tight_layout()
plt.show()
print("\n8B - PAIRPLOT")

pairplot_columns = [
    "Study_Hours_Per_Day",
    "Sleep_Hours_Per_Day",
    "Social_Hours_Per_Day",
    "Physical_Activity_Hours_Per_Day",
    "GPA"
]

sns.pairplot(
    df[pairplot_columns],
    diag_kind="hist"
)

plt.suptitle(
    "Pairwise Relationships Among Lifestyle and Academic Variables",
    y=1.02
)

plt.show()

print("\n8C - LIFESTYLE VARIABLES AND GPA")

lifestyle_columns = [
    "Study_Hours_Per_Day",
    "Sleep_Hours_Per_Day",
    "Social_Hours_Per_Day",
    "Physical_Activity_Hours_Per_Day",
    "Extracurricular_Hours_Per_Day"
]

# Create correlation table
lifestyle_gpa = (
    df[lifestyle_columns + ["GPA"]]
    .corr()["GPA"]
    .drop("GPA")
    .sort_values(ascending=False)
)

display(
    lifestyle_gpa.to_frame(
        name="Correlation_with_GPA"
    )
)


print("\n" + "=" * 70)
print("MULTIVARIATE ANALYSIS COMPLETED")
print("=" * 70)
```

```python

print("=" * 70)
print("STEP 9 - OUTLIER ANALYSIS")
print("=" * 70)

# Numerical variables for outlier analysis
outlier_columns = [
    "Study_Hours_Per_Day",
    "Extracurricular_Hours_Per_Day",
    "Sleep_Hours_Per_Day",
    "Social_Hours_Per_Day",
    "Physical_Activity_Hours_Per_Day",
    "GPA"
]
print("\n9A - BOXPLOTS FOR OUTLIER DETECTION")

for column in outlier_columns:
    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[column])

    plt.title(f"Outlier Analysis - {column}")
    plt.xlabel(column)

    plt.tight_layout()
    plt.show()
print("\n9B - IQR OUTLIER SUMMARY")

outlier_summary = []

for column in outlier_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Variable": column,
        "Q1": round(Q1, 2),
        "Q3": round(Q3, 2),
        "IQR": round(IQR, 2),
        "Lower_Bound": round(lower_bound, 2),
        "Upper_Bound": round(upper_bound, 2),
        "Outlier_Count": len(outliers)
    })


outlier_summary_df = pd.DataFrame(outlier_summary)

display(outlier_summary_df)
outlier_summary_df["Outlier_Percentage"] = (
    outlier_summary_df["Outlier_Count"]
    / len(df)
    * 100
).round(2)

print("\nOutlier Percentage:")

display(
    outlier_summary_df[
        ["Variable", "Outlier_Count", "Outlier_Percentage"]
    ]
)

print("\nOUTLIER HANDLING DECISION")
print("-" * 70)

print(
    "Potential outliers were identified using the IQR method."
)

print(
    "Outliers are retained because they may represent "
    "genuine differences in student lifestyle or academic performance."
)

print(
    "No observations are removed automatically unless there is "
    "evidence that they are data-entry errors or invalid values."
)


print("\n" + "=" * 70)
print("OUTLIER ANALYSIS COMPLETED")
print("=" * 70)
```
